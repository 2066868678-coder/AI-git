```python
from calendar import month

import  pandas as pd
pumpkins = pd.read_csv('US-pumpkins.csv')
pumpkins.head()
pumpkins.isnull().sum()
```

```python
columns_to_select = ['Package', 'Low Price', 'High Price', 'Date']
pumpkins = pumpkins.loc[:,columns_to_select]
print(pumpkins)
```

```python
price = (pumpkins["Low Price"] + pumpkins['High Price']) / 2
month = pd.DatetimeIndex(pumpkins['Date']).month
print(price)
print(month)
```

```python
new_pumpkins = pd.DataFrame({"月份":month, '包装方式':pumpkins['Package'], '最低价格': pumpkins['Low Price'],'最高价格': pumpkins['High Price'], '均价': price})
print(new_pumpkins)
```
