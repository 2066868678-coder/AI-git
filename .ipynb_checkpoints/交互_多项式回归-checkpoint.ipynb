# 🎮 交互式多项式回归
拖动滑块，曲线实时变化！

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error
from ipywidgets import interact, FloatSlider
%matplotlib inline

plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'SimSun']
plt.rcParams['axes.unicode_minus'] = False
```

## 🌡️ 温度 vs 用电量
拖动 degree 滑块，看曲线怎么变

```python
np.random.seed(42)
temp = np.array([-5, 0, 5, 10, 15, 20, 25, 30, 35, 38])
elec = 15 + 0.3 * (temp - 20)**2 + np.random.randn(10) * 3

x_smooth = np.linspace(-7, 42, 300)

def plot_poly(degree):
    poly = PolynomialFeatures(degree=degree, include_bias=False)
    temp_poly = poly.fit_transform(temp.reshape(-1,1))
    model = LinearRegression()
    model.fit(temp_poly, elec)
    pred = model.predict(temp_poly)
    
    x_sp = poly.fit_transform(x_smooth.reshape(-1,1))
    y_smooth = model.predict(x_sp)
    
    rmse = root_mean_squared_error(pred, elec)
    r2 = model.score(temp_poly, elec)
    
    plt.figure(figsize=(10,5))
    plt.scatter(temp, elec, color='#4D96FF', s=80, zorder=5, label='实际数据')
    plt.plot(x_smooth, y_smooth, color='#FF6B6B', linewidth=3, label=f'degree={degree}')
    plt.xlabel('温度 (\u2103)')
    plt.ylabel('用电量')
    plt.title(f'\ud83c\udfaf 多项式回归 degree={degree}  RMSE={rmse:.2f}  R\u00b2={r2:.3f}', fontsize=14)
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.ylim(0, 35)
    plt.show()

interact(plot_poly, degree=(1, 15, 1))
```

## 🌀 梯度下降动画
点一次按钮走一步，看模型怎么下山

```python
from ipywidgets import Button, Output
from IPython.display import display, clear_output

# 生成数据
np.random.seed(42)
X = np.random.randn(50, 1) * 2
y = 3 + 1.5 * X.ravel() + np.random.randn(50) * 0.5

# 初始化参数
w0, w1 = -2, -2
lr = 0.05
path = [(w0, w1)]

btn = Button(description='\u25b6 走一步')
out = Output()

@out.capture()
def step(b):
    global w0, w1
    # 预测
    y_pred = w0 + w1 * X.ravel()
    # 梯度
    grad_w0 = -2 * np.mean(y - y_pred)
    grad_w1 = -2 * np.mean((y - y_pred) * X.ravel())
    # 更新
    w0 -= lr * grad_w0
    w1 -= lr * grad_w1
    path.append((w0, w1))
    
    # 画图
    plt.figure(figsize=(10, 4))
    
    # 左图：拟合线
    plt.subplot(1, 2, 1)
    plt.scatter(X, y, alpha=0.6, color='#4D96FF')
    x_line = np.array([-4, 4])
    plt.plot(x_line, w0 + w1 * x_line, 'r-', linewidth=2)
    plt.xlim(-4, 4)
    plt.ylim(-2, 8)
    plt.title(f'y = {w0:.2f} + {w1:.2f}x')
    plt.grid(True, alpha=0.3)
    
    # 右图：损失下降路径
    plt.subplot(1, 2, 2)
    w0s = [p[0] for p in path]
    w1s = [p[1] for p in path]
    plt.plot(w0s, w1s, '.-', color='#FF6B6B', linewidth=2, markersize=8)
    plt.scatter(w0s[0], w1s[0], c='#FFD93D', s=100, label='起点', marker='o')
    plt.scatter(w0s[-1], w1s[-1], c='#00FF88', s=100, label='当前位置', marker='*')
    plt.xlabel('w0')
    plt.ylabel('w1')
    plt.title(f'\u4e0b\u5c71\u8def\u5f84  ({len(path)-1}\u6b65)')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

btn.on_click(step)
display(btn, out)
```
